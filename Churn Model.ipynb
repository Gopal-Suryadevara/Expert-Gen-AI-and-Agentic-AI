```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
x[:, 2] = le.fit_transform(x[:, 2])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
x = np.array(ct.fit_transform(x))
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x = sc.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

# ANN MODEL BUILDING ARCHITECTURE

```python
ann = tf.keras.models.Sequential() #establishing the layer one after the other
```

```python
ann.add(tf.keras.layers.Dense(units=6)) #input layer
```

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu')) #input layer and first hidden layer
```

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) #Output layer
```

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
ann.fit(x_train, y_train, batch_size = 32, epochs = 50)
```
